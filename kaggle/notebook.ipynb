# BHMarketer TechBullion FAQ Generator: Search Walkthrough

This notebook runs the [BHMarketer TechBullion FAQ Generator](https://github.com/bhmarketer-ai/BHMarketer-buy-techbullion-press-release-faq-generator) on 40 real-world questions about buying a TechBullion press release, checks each one finds the expected FAQ, and generates the FAQ with FAQPage schema. Built by [BHMarketer.ai](https://bhmarketer.ai/products/techbullion/).

## 1. Load the tool and the datasets

```python
import os, sys
import pandas as pd

BASES = ["/kaggle/input/techbullion-faq-dataset", "..", "."]
base = next(b for b in BASES if os.path.exists(os.path.join(b, "techbullion_faq.py")))
data = base if os.path.exists(os.path.join(base, "techbullion_faqs.csv")) else os.path.join(base, "dataset")
sys.path.insert(0, base)
import techbullion_faq as tb

faqs = pd.read_csv(os.path.join(data, "techbullion_faqs.csv"))
bench = pd.read_csv(os.path.join(data, "techbullion_search_benchmarks.csv"))
print(len(faqs), "FAQs |", len(bench), "benchmark questions")
```

## 2. FAQs per category

```python
faqs.groupby("category").size().rename("faqs").to_frame()
```

## 3. Run every benchmark question

```python
rows = []
for r in bench.itertuples():
    a = tb.ask(r.query)
    rows.append({"query": r.query, "found": a["id"], "expected": r.expected_id,
                 "confidence": a["confidence"], "band": a["band"], "pass": a["id"] == r.expected_id})
results = pd.DataFrame(rows)
print(f"{results['pass'].sum()}/{len(results)} benchmark questions matched")
results
```

```python
results.groupby("band").size().rename("questions").to_frame()
```

## 4. Ask your own question

```python
for r in tb.search("can i launch a presale campaign on techbullion", 3):
    print(f"Q{r['id']} [{r['confidence']}] {r['question']}")

best = tb.ask("how much does a techbullion press release cost")
print("\n", best["answer"])
```

## 5. Generate the FAQ with FAQPage schema

```python
print(tb.export("jsonld", "buying")[:1200])
```
